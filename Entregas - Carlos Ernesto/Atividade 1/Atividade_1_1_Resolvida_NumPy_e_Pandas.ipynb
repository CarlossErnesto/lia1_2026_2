## ✅ Solução do Desafio 1

Os dados de vendas são recriados para que a solução funcione de forma independente. Em seguida, os registros são agrupados para identificar os maiores resultados e calcular as médias solicitadas.


```python
# Importando as bibliotecas necessárias
import numpy as np
import pandas as pd

# Recriando os dados de vendas utilizados na atividade
np.random.seed(42)

dados_vendas = {
    'Vendedor': ['Ana', 'João', 'Pedro', 'Ana', 'João', 'Pedro', 'Ana', 'João', 'Pedro', 'Ana'] * 3,
    'Produto': ['Notebook', 'Mouse', 'Teclado', 'Monitor', 'Headset'] * 6,
    'Regiao': ['Norte', 'Sul', 'Nordeste', 'Centro-Oeste', 'Sudeste'] * 6,
    'Vendas': np.random.randint(1000, 10000, 30),
    'Quantidade': np.random.randint(1, 20, 30)
}

df_vendas_completo = pd.DataFrame(dados_vendas)

# Criando períodos para analisar a evolução das vendas por região
df_vendas_completo['Periodo'] = np.repeat(np.arange(1, 7), 5)

# Agrupando os dados para responder às perguntas
numero_vendas_por_vendedor = df_vendas_completo.groupby('Vendedor').size()
faturamento_por_vendedor = df_vendas_completo.groupby('Vendedor')['Vendas'].sum()
faturamento_por_regiao = df_vendas_completo.groupby('Regiao')['Vendas'].sum()
quantidade_por_produto = df_vendas_completo.groupby('Produto')['Quantidade'].sum()
media_vendas_por_vendedor = df_vendas_completo.groupby('Vendedor')['Vendas'].mean()

# Identificando os maiores resultados
vendedor_destaque = numero_vendas_por_vendedor.idxmax()
regiao_destaque = faturamento_por_regiao.idxmax()
produto_destaque = quantidade_por_produto.idxmax()

# Formatando valores monetários no padrão brasileiro
def moeda_br(valor):
    return f"R$ {valor:,.2f}".replace(',', 'X').replace('.', ',').replace('X', '.')

print("📊 RESULTADOS DA ANÁLISE DE VENDAS")
print(f"1️⃣ Vendedor com maior número de vendas: {vendedor_destaque} - {numero_vendas_por_vendedor.max()} vendas")
print(f"2️⃣ Região com maior faturamento: {regiao_destaque} - {moeda_br(faturamento_por_regiao.max())}")
print(f"3️⃣ Produto mais vendido em quantidade: {produto_destaque} - {quantidade_por_produto.max()} unidades")
print("4️⃣ Média de vendas por vendedor:")
for vendedor, media in media_vendas_por_vendedor.items():
    print(f"   • {vendedor}: {moeda_br(media)}")
```

## ✅ Solução do Desafio 2

Foi criado um conjunto de dados com duplicatas, valores ausentes e textos despadronizados. A solução remove as inconsistências e acrescenta colunas calculadas a partir dos dados tratados.


```python
# Criando um DataFrame com duplicatas, valores ausentes e strings despadronizadas
dados_problematicos = {
    'Produto': [' notebook ', 'MOUSE', 'Teclado', 'MOUSE', 'monitor', 'Headset'],
    'Categoria': [' informática', 'ACESSÓRIOS ', 'Acessórios', 'ACESSÓRIOS ', None, 'acessórios'],
    'Quantidade': [2, 5, None, 5, 3, 4],
    'Preco_Unitario': [3500.00, 80.00, 150.00, 80.00, None, 250.00]
}

df_problematico = pd.DataFrame(dados_problematicos)
print("📋 DataFrame original:")
display(df_problematico)

# Removendo duplicatas
df_produtos_limpo = df_problematico.drop_duplicates().copy()

# Tratando valores ausentes
df_produtos_limpo['Categoria'] = df_produtos_limpo['Categoria'].fillna('Informática')
df_produtos_limpo['Quantidade'] = df_produtos_limpo['Quantidade'].fillna(
    df_produtos_limpo['Quantidade'].median()
)
df_produtos_limpo['Preco_Unitario'] = df_produtos_limpo['Preco_Unitario'].fillna(
    df_produtos_limpo['Preco_Unitario'].median()
)

# Padronizando as strings
df_produtos_limpo['Produto'] = df_produtos_limpo['Produto'].str.strip().str.title()
df_produtos_limpo['Categoria'] = df_produtos_limpo['Categoria'].str.strip().str.title()

# Criando novas colunas derivadas
df_produtos_limpo['Valor_Total'] = (
    df_produtos_limpo['Quantidade'] * df_produtos_limpo['Preco_Unitario']
)
df_produtos_limpo['Faixa_Valor'] = pd.cut(
    df_produtos_limpo['Valor_Total'],
    bins=[0, 500, 2000, float('inf')],
    labels=['Baixo', 'Médio', 'Alto']
)

print("✅ DataFrame após a limpeza:")
display(df_produtos_limpo)
print(f"🧹 Valores ausentes restantes: {df_produtos_limpo.isnull().sum().sum()}")
```

## ✅ Solução do Desafio 3

As informações obtidas nos desafios anteriores são apresentadas em três gráficos: comparação entre vendedores, evolução das vendas por região e distribuição dos valores. Para representar a tendência regional, cada bloco de cinco registros foi considerado um período de vendas.


```python
# Importando as bibliotecas de visualização
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
fig, eixos = plt.subplots(1, 3, figsize=(20, 6))

# Comparação entre vendedores
faturamento_por_vendedor.sort_values().plot(
    kind='barh', color='#2E86AB', ax=eixos[0]
)
eixos[0].set_title('📊 Faturamento por Vendedor')
eixos[0].set_xlabel('Faturamento (R$)')
eixos[0].set_ylabel('Vendedor')

# Tendência das vendas por região ao longo dos períodos
sns.lineplot(
    data=df_vendas_completo,
    x='Periodo', y='Vendas', hue='Regiao',
    marker='o', linewidth=2, ax=eixos[1]
)
eixos[1].set_title('🌎 Tendência de Vendas por Região')
eixos[1].set_xlabel('Período')
eixos[1].set_ylabel('Valor das Vendas (R$)')
eixos[1].legend(title='Região', fontsize=8)

# Distribuição dos valores de vendas
sns.histplot(
    data=df_vendas_completo, x='Vendas', bins=10,
    kde=True, color='#A23B72', ax=eixos[2]
)
eixos[2].set_title('📈 Distribuição dos Valores de Vendas')
eixos[2].set_xlabel('Valor das Vendas (R$)')
eixos[2].set_ylabel('Frequência')

plt.tight_layout()
plt.show()
```
